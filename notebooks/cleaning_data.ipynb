### Питання для дослідження:
Які відмінності в рівні заробітної плати працівників в Україні спостерігалися у 2016 та 2020 роках залежно від статі, рівня освіти, вікової категорії, стажу роботи, професії та виду економічної діяльності?

**Гіпотези:**

- **Освіта:** рівень заробітної плати відрізняється залежно від рівня освіти та загалом є вищим серед працівників із вищим рівнем освіти. 
- **Вік:** рівень заробітної плати відрізняється між віковими групами та зростає до певної вікової категорії, після чого може стабілізуватися або знижуватися. 
- **Стаж:** працівники з більшим стажем мають вищий рівень заробітної плати.
- **Професія:** рівень зарплати суттєво відрізняється між професійними групами.
- **Вид діяльності:** рівень заробітної плати суттєво відрізняється між видами економічної діяльності. 
- **Стать:** рівень заробітної плати відрізняється між чоловіками та жінками, причому середня заробітна плата чоловіків є вищою. 



### 3. Визначення даних для аналізу

Згідно з **питанням дослідження** та сформованими **гіпотезами**, визначаємо, які дані з датасету використовуємо для подальшого аналізу:

- **Територіальний розріз** — залишаємо значення **«Україна»**, оскільки аналіз проводиться на рівні всієї країни без розбиття за регіонами.

- **Категорія розрізу** — залишаємо такі категорії:
  - **«Професії та професійні групи»**
  - **«Вікова група»**
  - **«Вид економічної діяльності»**
  - **«Рівень освіти»**
  - **«Стаж роботи»**

- **Стать** — залишаємо значення **"Чоловіки"** та **"Жінки"** для дослідження рівня заробітної плати по гендерному розподілу.

```python
# Перевіримо дані по визначеним розрізам для аналізу. Зробимо копію датасету, де залишимо дані, потрібні для дослідження.

df_clean = df.copy()

categories = [
    'Професії та професійні групи',
    'Вікова група',
    'Вид економічної діяльності',
    'Рівень освіти',
    'Стаж роботи'
]

df_clean = df[
    (df['Територіальний розріз'] == 'Україна') &
    (df['Стать'].isin(['Жінки', 'Чоловіки'])) &
    (df['Категорія розрізу'].isin(categories))
].copy()

```

```python
# Перевірка даних після фільтрації
print('Кількість рядків після фільтрації:', len(df_clean))

print('\nТериторіальний розріз:')
print(df_clean['Територіальний розріз'].value_counts())

print('\nСтать:')
print(df_clean['Стать'].value_counts())

print('\nКатегорія розрізу:')
print(df_clean['Категорія розрізу'].value_counts())
```

```python
# Перевіряємо пусті значення у даних
print(df_clean.isna().sum())
```

```python
# Видаляємо повністю пусті колонки, які не несуть цінності для дослідження

df_clean = df_clean.drop(columns=[
    '2017',
    '2018',
    '2019',
    'Примітки набору даних'
])
```

```python
# Перевіряємо дані після видалення пустих колонок

print(df_clean.isna().sum())
```

```python
# Бачимо з попереднього результату, що є пропуски в колонках 2016 та 2020, що потрібні нам для дослідження.
# Тому робимо перевірку пустих значень з виводом додаткової інформації по цим полям, чи не зазначено причину пропуску
df_clean[df_clean['2016'].isna() | df_clean['2020'].isna()][[
    'Категорія розрізу',
    'Розріз',
    'Спеціальне значення даних',
    'Примітки часового ряду',
    'Примітки спостереження',
    '2016',
    '2020'
]]
```

```python
# Опрацювання відсутніх даних. Створюємо окремий статус для пропусків

df_clean['Причина пропуску'] = 'Немає пропусків'

# 1. Явищ не було
df_clean.loc[
    df_clean['Спеціальне значення даних'].eq('Явищ не було'),
    'Причина пропуску'
] = 'Явищ не було'

# 2. Інформація відсутня
df_clean.loc[
    df_clean['Спеціальне значення даних'].eq('Інформація відсутня.') |
    df_clean['Примітки спостереження'].eq('Інформація відсутня.'),
    'Причина пропуску'
] = 'Інформація відсутня'

# 3. Пропуск є, але причина не зазначена
df_clean.loc[
    (
        df_clean['2016'].isna() |
        df_clean['2020'].isna()
    ) &
    (df_clean['Причина пропуску'] == 'Немає пропусків'),
    'Причина пропуску'
] = 'Причина не зазначена'
```

```python
# Перевіряємо інформацію після присвоєння статусу для пропущених значень щоб дізнатися для скількох з них причина пропуску була вказана
print(
    df_clean.loc[
        df_clean['2016'].isna() | df_clean['2020'].isna(),
        'Причина пропуску'
    ].value_counts()
)
```

Пропуски не замінюємо на 0.
NaN залишаємо як NaN, а причину зберігаємо окремим полем **Причина пропуску**.

## 4. Перевіряємо гранулярність

**Бажана гранулярність:**

Категорія розрізу → Розріз → Стать → одне спостереження за 2016 та 2020 роки

```python
granularity_cols = [
    'Категорія розрізу',
    'Розріз',
    'Стать'
]
```

```python
# Перевіряємо наявність дублікатів

duplicates = df_clean[
    df_clean.duplicated(
        subset=granularity_cols,
        keep=False
    )
]

print('Кількість дубльованих рядків:', len(duplicates))
```

```python
# Перевіряємо, які саме дублікати є в датасеті
duplicate_counts = (
    df_clean
    .groupby(
        ['Категорія розрізу', 'Розріз', 'Стать'],
        dropna=False
    )
    .size()
    .reset_index(name='Кількість')
)

print(
    duplicate_counts[
        duplicate_counts['Кількість'] > 1
    ].to_string(index=False)
)
```

```python
# Детальніше дослідження дублікатів для порівняння
print(
    duplicates[
        [
            'Стать',
            'Категорія розрізу',
            'Розріз',
            '2016',
            '2020',
            'Спеціальне значення даних',
            'Примітки часового ряду',
            'Примітки спостереження'
        ]
    ]
    .sort_values(
        ['Розріз', 'Стать']
    )
    .to_string(index=False)
)
```

## Висновок щодо дублікатів

**Під час перевірки дублікатів виявлено два типи неоднозначних спостережень, для яких прийнято рішення про виключення з аналізу:**

- **«Інші професійні назви робіт кваліфікованих робітників з інструментом»**

- один рядок містить дані про середньомісячну заробітну плату;
- другий — дані про середню заробітну плату за одну оплачену годину.

Погодинне спостереження видаляємо, оскільки в межах цього дослідження аналізується саме середньомісячна заробітна плата.

- **«Керівники підприємств, установ та організацій»**

два спостереження мають однакові **Категорія розрізу**, **Розріз**, **Стать** та **Примітки**;
при цьому значення заробітної плати за 2016 та 2020 роки відрізняються;
у даних відсутнє пояснення причини такої розбіжності.

Через недостатність інформації неможливо достовірно визначити, яке з двох значень є коректним. Тому обидва спостереження виключено з аналізу, щоб уникнути необґрунтованого вибору даних та потенційного спотворення результатів дослідження.

```python
# 1. Видаляємо погодинні спостереження
mask_hourly = (
    (df_clean['Категорія розрізу'] == 'Професії та професійні групи') &
    (df_clean['Розріз'] == 'Інші професійні назви робіт кваліфікованих робітників з інструментом') &
    df_clean['Примітки часового ряду'].str.contains(
        'за одну оплачену годину',
        na=False
    )
)

# 2. Видаляємо неоднозначні спостереження щодо керівників
mask_ambiguous = (
    (df_clean['Категорія розрізу'] == 'Професії та професійні групи') &
    (df_clean['Розріз'].str.strip() == 'Керівники підприємств, установ та організацій')
)

# Видаляємо визначені дублікати
df_clean = df_clean.loc[
    ~(mask_hourly | mask_ambiguous)
].copy()

print('Кількість рядків після видалення:', len(df_clean))
```

```python
# Перевірка пропусків після очищення
print('Пропуски після очищення:')
print(
    df_clean.isna()
    .sum()
    .sort_values(ascending=False)
)
```

```python
# Перевірка пропусків, де значення відсутні в обох роках досліджень, так як ці дані не несуть жодної інформації про рівень зарплати
df_clean.loc[
    df_clean['2016'].isna() & df_clean['2020'].isna(),
    [
        'Стать',
        'Категорія розрізу',
        'Розріз',
        '2016',
        '2020',
        'Причина пропуску'
    ]
].to_string(index=False)
```

```python
# Видаляємо рядки без значень за обидва роки досліджень
df_clean = df_clean.dropna(
    subset=['2016', '2020'],
    how='all'
)

print('Кількість рядків після видалення:', len(df_clean))
```

```python
# Так як для нашого дослідження потрібні повні дані за 2016 і 2020 для порівняння, для датасету з яким будемо працювати надалі,
# видалимо рядки, де є пропущені дані

df_tableau = df_clean.copy()

# Видаляємо рядки, де відсутнє значення за 2016 або 2020
df_tableau = df_tableau.dropna(
    subset=['2016', '2020']
).copy()

print('Кількість рядків:', len(df_tableau))
print('Кількість колонок:', len(df_tableau.columns))
```

```python
df_tableau_long = df_tableau.melt(
    id_vars=[
        'Показник',
        'Період',
        'Стать',
        'Територіальний розріз',
        'Категорія розрізу',
        'Розріз',
        'Періодичність',
        'Одиниця виміру',
        'Масштабування',
        'Кількість десяткових знаків',
        'Спеціальне значення даних',
        'Примітки часового ряду',
        'Примітки спостереження',
        'Причина пропуску'
    ],
    value_vars=['2016', '2020'],
    var_name='Рік',
    value_name='Заробітна плата'
)
```

```python
# Вивантажуємо дані для подальшого опрацювання в tableau

df_tableau_long.to_csv(
   'salary_research_tableau.csv',
   index=False,
   encoding='utf-8-sig',
   sep=';'
)
```
